# Mutual Fund Analytics: Day 1 - Data Ingestion & EDA
This notebook provides interactive exploration of the 10 mutual fund datasets, AMFI code integrity checks, and live NAV ingestion from mfapi.in.

```
import os
import glob
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

%matplotlib inline
plt.style.use('seaborn-v0_8-whitegrid')
```

```
# Load Fund Master and NAV History
fm_df = pd.read_csv('../data/raw/01_fund_master.csv')
nav_df = pd.read_csv('../data/raw/02_nav_history.csv')

print(f'Fund Master Shape: {fm_df.shape}')
print(f'NAV History Shape: {nav_df.shape}')
fm_df.head()
```

```
# Cross-validation of AMFI Codes
fm_codes = set(fm_df['amfi_code'].unique())
nav_codes = set(nav_df['amfi_code'].unique())

print(f'Total Unique AMFI codes in Fund Master: {len(fm_codes)}')
print(f'Total Unique AMFI codes in NAV History: {len(nav_codes)}')
print(f'Are all fund_master codes present in nav_history? {fm_codes.issubset(nav_codes)}')
```

```
# Fund Houses and Scheme Counts
plt.figure(figsize=(10, 5))
fm_df['fund_house'].value_counts().plot(kind='barh', color='teal')
plt.title('Number of Schemes by Fund House')
plt.xlabel('Scheme Count')
plt.ylabel('Fund House')
plt.tight_layout()
plt.show()
```
